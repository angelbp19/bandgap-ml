```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/MasterAI-EAM/Darwin/main/dataset/matbench_expt_gap/matbench_expt_gap.csv"
df = pd.read_csv(url).rename(columns={"band gap": "gap"})

print(df.shape)
print(df.head(10))
print(df["gap"].describe())
print("Fraction with gap = 0:", (df["gap"] == 0).mean())
```
